```python
diamonds_clean = diamonds.drop_duplicates()
```

```python
diamonds_clean.shape
```

```python
(diamonds_clean[["x", "y", "z"]] == 0).sum()
```

```python
diamonds_clean[
    (diamonds_clean["x"] == 0) |
    (diamonds_clean["y"] == 0) |
    (diamonds_clean["z"] == 0)
]
```

```python
diamonds_clean = diamonds_clean[
    (diamonds_clean["x"] > 0) &
    (diamonds_clean["y"] > 0) &
    (diamonds_clean["z"] > 0)
]
```

```python
diamonds_clean.shape
```

```python
sns.scatterplot(data=diamonds_clean, x="carat", y="price")
```

```python
sns.boxplot(data=diamonds_clean, x="cut", y="price")
```

```python
sns.heatmap(
    diamonds_clean[["carat", "depth", "table", "price", "x", "y", "z"]].corr(),
    annot=True,
    cmap="coolwarm"
)
```

```python
sns.histplot(data=diamonds_clean, x="price", bins=30, kde=True)
```

# تحلیل داده‌های الماس (Diamonds Data Analysis)

## ۱. معرفی مجموعه داده
مجموعه داده Diamonds شامل ۵۳٬۹۴۰ ردیف و ۱۰ ستون است. این داده‌ها اطلاعاتی درباره وزن، رنگ، شفافیت، کیفیت برش، ابعاد و قیمت الماس‌ها ارائه می‌دهند.

## ۲. پاک‌سازی داده‌ها
- هیچ مقدار گمشده‌ای در ستون‌ها پیدا نشد.
- ۱۴۶ ردیف تکراری حذف شد.
- ۱۹ ردیف که حداقل یکی از ابعاد آن‌ها (x، y یا z) صفر بود، حذف شد.

پس از پاک‌سازی، مجموعه داده شامل **۵۳٬۷۷۵ ردیف و ۱۰ ستون** شد.

## ۳. تحلیل اکتشافی داده‌ها
- گروه Ideal بیشترین فراوانی را در کیفیت برش داشت.
- رنگ G پرتکرارترین رنگ در مجموعه داده بود.
- درجه شفافیت SI1 بیشترین فراوانی را داشت.
- بین وزن الماس (carat) و قیمت آن رابطه مثبت و قوی وجود داشت؛ ضریب همبستگی این دو برابر ۰٫۹۲ بود.
- توزیع قیمت الماس‌ها چولگی به راست داشت؛ یعنی بیشتر الماس‌ها قیمت پایین‌تری داشتند و تعداد کمی از آن‌ها بسیار گران‌قیمت بودند.
